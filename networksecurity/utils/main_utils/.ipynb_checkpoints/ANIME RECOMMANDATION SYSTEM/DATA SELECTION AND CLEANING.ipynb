# **starting encoding**

```python
df_knn.shape
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MultiLabelBinarizer
from sklearn.preprocessing import StandardScaler
```

```python
x=df_knn.copy()
```

```python
ohe = OneHotEncoder(sparse_output=False)
scale=StandardScaler()

ct = ColumnTransformer(
    transformers=[
        ('ohe', ohe, ['Type', 'Source', 'Rating']),
        ('scale', scale, ['Episodes'])
    ],
    remainder='passthrough'
)
```

```python
X_transformed = ct.fit_transform(x)
```

```python
X_transformed_df = pd.DataFrame(
    X_transformed,
    columns=ct.get_feature_names_out(),
    index=x.index
)
```

```python
X_transformed_df.head(1)
```

```python
X_transformed_df.columns = (
    X_transformed_df.columns
    .str.replace('ohe__', '', regex=False)
    .str.replace('scale__', '', regex=False)
    .str.replace('remainder__', '', regex=False)
)
```

```python
X_transformed_df.head(1)
```

```python
X_transformed_df['Genres'] = X_transformed_df['Genres'].apply(
    lambda x: [genre.strip() for genre in x.split(',')]
)
```

```python
X_transformed_df['Studios'] = X_transformed_df['Studios'].apply(
    lambda x: [studios.strip() for studios in x.split(',')]
)
```

```python
X_transformed_df['Genres']
```

```python
X_transformed_df['Studios']
```

```python
from sklearn.preprocessing import MultiLabelBinarizer
mlb_genres = MultiLabelBinarizer()
mlb_studios = MultiLabelBinarizer()
encoded_genres = mlb_genres.fit_transform(X_transformed_df['Genres'])
encoded_studios = mlb_studios.fit_transform(X_transformed_df['Studios'])
```

```python
genres_df = pd.DataFrame(
    encoded_genres,
    columns=mlb_genres.classes_,
    index=X_transformed_df.index
)
```

```python
studios_df = pd.DataFrame(
    encoded_studios,
    columns=mlb_studios.classes_,
    index=X_transformed_df.index
)
```

```python
X_transformed_df=X_transformed_df.drop(columns=['Genres','Studios'])
```

```python
X_final = pd.concat(
    [X_transformed_df, genres_df, studios_df],
    axis=1
)
```

```python
X_transformed_df = X_transformed_df.apply(pd.to_numeric)
```

```python
X_final = pd.concat(
    [X_transformed_df, genres_df, studios_df],
    axis=1
)
```

## Training Model using **KNN**

```python
from sklearn.neighbors import NearestNeighbors
knn=NearestNeighbors(n_neighbors=11,metric='cosine')
knn.fit(X_final)
```

```python
df_ui=pd.read_csv(r'/content/drive/MyDrive/ANIME RECOMMANDATION SYSTEM/anime_ui.csv')
```

# Testing on **Data**

```python
anime_name=input('enter anime name: ')
anime_index = df_ui[df_ui['Name'] == anime_name].index[0]

distances, indices = knn.kneighbors(
    X_final.iloc[[anime_index]]
)
df_ui.iloc[indices[0]][['Name', 'Genres', 'Score']]
```

```python
def recommend_anime(anime_name=input('enter anime name')):
  anime_index = df_ui[df_ui['Name'] == anime_name].index[0]

  distances, indices = knn.kneighbors(
      X_final.iloc[[anime_index]]
  )
  return df_ui.iloc[indices[0]][['Name', 'Genres', 'Score']]
```
