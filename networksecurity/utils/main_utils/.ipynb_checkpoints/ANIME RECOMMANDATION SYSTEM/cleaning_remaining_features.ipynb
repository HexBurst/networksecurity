```python
#column need to be stripped of spaces
strip_space=['Name','English name','Other name','Synopsis','Aired','Duration','Rank','Status','Premiered','Producers','Licensors']
for col in strip_space:
  df[col]=df[col].str.strip()
```

```python
for col in strip_space:
    print(col, (df[col] != df[col].str.strip()).sum())
```

```python
df['Rank'] = pd.to_numeric(df['Rank'], errors='coerce')
```

```python
df['Rank'].isna().sum()
```

```python
df['Scored By'] = pd.to_numeric(df['Scored By'], errors='coerce')
```

```python
numeric_cols = [
    'Score',
    'Popularity',
    'Favorites',
    'Scored By',
    'Members'
]
for col in numeric_cols:
    print(col, (df[col].isna().sum()))
```

```python
text_cols = [
    'Name',
    'English name',
    'Other name',
    'Synopsis',
    'Aired',
    'Premiered',
    'Status',
    'Producers',
    'Licensors',
    'Duration'
]

for col in text_cols:
    df[col] = df[col].apply(
        lambda x: 'Unknown' if isinstance(x, str) and x.lower() == 'unknown' else x
    )
```

```python
for col in text_cols:
    print(f"{col}: {(df[col] == 'Unknown').sum()}")
```

```python
df_knn=pd.read_csv(r'/content/drive/MyDrive/ANIME RECOMMANDATION SYSTEM/anime_features_cleaned.csv')
```

```python
df_ui=pd.concat([df,df_knn],axis=1)
```
